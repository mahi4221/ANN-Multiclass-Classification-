```python
# ============================================================
# FASHION-MNIST IMAGE CLASSIFICATION USING ANN
# Multi-Class Classification
# 10 Classes: 0 - 9
# ============================================================


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout


# ============================================================
# 2. LOAD DATASET
# ============================================================

train_file = r"C:\Users\Maruthi B\Downloads\archive (3) (1)\fashion-mnist_train.csv"
test_file = r"C:\Users\Maruthi B\Downloads\archive (3) (1)\fashion-mnist_test.csv"

train_df = pd.read_csv(train_file)
test_df = pd.read_csv(test_file)

print("Training Dataset Shape:", train_df.shape)
print("Testing Dataset Shape:", test_df.shape)

print("\nFirst 5 Training Rows:")
print(train_df.head())


# ============================================================
# 3. DATASET INFORMATION
# ============================================================

print("\nTraining Dataset Information:")
print(train_df.info())

print("\nMissing Values in Training Data:")
print(train_df.isnull().sum().sum())

print("\nMissing Values in Testing Data:")
print(test_df.isnull().sum().sum())

print("\nDuplicate Rows in Training Data:")
print(train_df.duplicated().sum())

print("\nDuplicate Rows in Testing Data:")
print(test_df.duplicated().sum())


# ============================================================
# 4. REMOVE DUPLICATES
# ============================================================

train_df = train_df.drop_duplicates()
test_df = test_df.drop_duplicates()

print("\nTraining Shape After Removing Duplicates:")
print(train_df.shape)

print("\nTesting Shape After Removing Duplicates:")
print(test_df.shape)


# ============================================================
# 5. SEPARATE FEATURES AND TARGET
# ============================================================

# label is the target column
# All remaining columns are pixel features

X_train = train_df.drop(columns=["label"])
y_train = train_df["label"]

X_test = test_df.drop(columns=["label"])
y_test = test_df["label"]


print("\nTraining Features Shape:")
print(X_train.shape)

print("\nTraining Target Shape:")
print(y_train.shape)

print("\nTesting Features Shape:")
print(X_test.shape)

print("\nTesting Target Shape:")
print(y_test.shape)


# ============================================================
# 6. CHECK TARGET CLASSES
# ============================================================

print("\nTarget Classes:")
print(sorted(y_train.unique()))

print("\nNumber of Classes:")
print(y_train.nunique())

print("\nClass Distribution:")
print(y_train.value_counts().sort_index())


# ============================================================
# 7. VISUALIZE CLASS DISTRIBUTION
# ============================================================

plt.figure(figsize=(8, 5))

sns.countplot(
    x=y_train
)

plt.title("Fashion-MNIST Class Distribution")

plt.xlabel("Class")

plt.ylabel("Number of Images")

plt.show()


# ============================================================
# 8. DISPLAY SAMPLE IMAGES
# ============================================================

class_names = [
    "T-shirt/top",
    "Trouser",
    "Pullover",
    "Dress",
    "Coat",
    "Sandal",
    "Shirt",
    "Sneaker",
    "Bag",
    "Ankle boot"
]


plt.figure(figsize=(10, 6))

for i in range(10):

    image = X_train.iloc[i].values.reshape(28, 28)

    plt.subplot(2, 5, i + 1)

    plt.imshow(
        image,
        cmap="gray"
    )

    plt.title(
        class_names[y_train.iloc[i]]
    )

    plt.axis("off")

plt.tight_layout()

plt.show()


# ============================================================
# 9. NORMALIZE PIXEL VALUES
# ============================================================

# Pixel values are between 0 and 255.
# Convert them into values between 0 and 1.

X_train = X_train / 255.0
X_test = X_test / 255.0

print("\nPixel Values After Normalization:")

print(X_train.iloc[:5, :10])


# ============================================================
# 10. CONVERT DATA TO NUMPY ARRAYS
# ============================================================

X_train = X_train.values
X_test = X_test.values

y_train = y_train.values
y_test = y_test.values


print("\nFinal Training Feature Shape:")
print(X_train.shape)

print("\nFinal Testing Feature Shape:")
print(X_test.shape)


# ============================================================
# 11. BUILD ANN MODEL
# ============================================================

model = Sequential()


# Input Layer + First Hidden Layer
model.add(
    Dense(
        256,
        activation="relu",
        input_shape=(784,)
    )
)


# Dropout Layer
model.add(
    Dropout(0.2)
)


# Second Hidden Layer
model.add(
    Dense(
        128,
        activation="relu"
    )
)


# Dropout Layer
model.add(
    Dropout(0.2)
)


# Third Hidden Layer
model.add(
    Dense(
        64,
        activation="relu"
    )
)


# Output Layer
# 10 neurons because Fashion-MNIST has 10 classes
model.add(
    Dense(
        10,
        activation="softmax"
    )
)


# ============================================================
# 12. DISPLAY ANN ARCHITECTURE
# ============================================================

print("\nANN Model Summary:")

model.summary()


# ============================================================
# 13. COMPILE ANN MODEL
# ============================================================

model.compile(

    optimizer="adam",

    loss="sparse_categorical_crossentropy",

    metrics=["accuracy"]
)


# ============================================================
# 14. TRAIN ANN MODEL
# ============================================================

history = model.fit(

    X_train,

    y_train,

    epochs=30,

    batch_size=64,

    validation_split=0.20,

    verbose=1
)


# ============================================================
# 15. EVALUATE MODEL
# ============================================================

test_loss, test_accuracy = model.evaluate(

    X_test,

    y_test,

    verbose=0
)


print("\n============================================")
print("MODEL EVALUATION")
print("============================================")

print(
    "Test Loss:",
    test_loss
)

print(
    "Test Accuracy:",
    test_accuracy
)


# ============================================================
# 16. MAKE PREDICTIONS
# ============================================================

# Predict probability for each of the 10 classes

y_pred_probability = model.predict(
    X_test
)


# Select class with highest probability

y_pred = np.argmax(
    y_pred_probability,
    axis=1
)


print("\nPredicted Classes:")
print(y_pred[:20])


# ============================================================
# 17. ACCURACY SCORE
# ============================================================

accuracy = accuracy_score(

    y_test,

    y_pred
)


print("\nAccuracy Score:")

print(accuracy)


# ============================================================
# 18. CLASSIFICATION REPORT
# ============================================================

print("\n============================================")
print("CLASSIFICATION REPORT")
print("============================================")

print(

    classification_report(

        y_test,

        y_pred,

        target_names=class_names

    )

)


# ============================================================
# 19. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(

    y_test,

    y_pred

)


print("\nConfusion Matrix:")

print(cm)


# ============================================================
# 20. VISUALIZE CONFUSION MATRIX
# ============================================================

plt.figure(figsize=(10, 8))

sns.heatmap(

    cm,

    annot=True,

    fmt="d",

    cmap="Blues",

    xticklabels=class_names,

    yticklabels=class_names

)

plt.title(
    "Fashion-MNIST ANN Confusion Matrix"
)

plt.xlabel(
    "Predicted Class"
)

plt.ylabel(
    "Actual Class"
)

plt.xticks(rotation=45)

plt.yticks(rotation=0)

plt.tight_layout()

plt.show()


# ============================================================
# 21. TRAINING VS VALIDATION ACCURACY
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(

    history.history["accuracy"],

    label="Training Accuracy"

)

plt.plot(

    history.history["val_accuracy"],

    label="Validation Accuracy"

)

plt.title(
    "Training vs Validation Accuracy"
)

plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Accuracy"
)

plt.legend()

plt.show()


# ============================================================
# 22. TRAINING VS VALIDATION LOSS
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(

    history.history["loss"],

    label="Training Loss"

)

plt.plot(

    history.history["val_loss"],

    label="Validation Loss"

)

plt.title(
    "Training vs Validation Loss"
)

plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Loss"
)

plt.legend()

plt.show()


# ============================================================
# 23. SAMPLE PREDICTIONS
# ============================================================

print("\n============================================")
print("SAMPLE PREDICTIONS")
print("============================================")


results = pd.DataFrame({

    "Actual": y_test[:20],

    "Predicted": y_pred[:20],

    "Probability": np.max(
        y_pred_probability[:20],
        axis=1
    )

})


print(results)


# ============================================================
# 24. DISPLAY SAMPLE PREDICTIONS
# ============================================================

plt.figure(figsize=(12, 8))

for i in range(10):

    image = X_test[i].reshape(28, 28)

    plt.subplot(2, 5, i + 1)

    plt.imshow(
        image,
        cmap="gray"
    )

    actual_class = class_names[y_test[i]]

    predicted_class = class_names[y_pred[i]]

    plt.title(
        f"Actual: {actual_class}\n"
        f"Predicted: {predicted_class}"
    )

    plt.axis("off")


plt.tight_layout()

plt.show()


# ============================================================
# 25. FINAL RESULTS
# ============================================================

print("\n============================================")
print("FINAL ANN RESULTS")
print("============================================")

print(
    f"Test Accuracy : {test_accuracy:.4f}"
)

print(
    f"Test Loss     : {test_loss:.4f}"
)

print(
    f"Accuracy Score: {accuracy:.4f}"
)


# ============================================================
# 26. EXTERNAL IMAGE PREDICTION
# ============================================================

print("\n============================================")
print("EXTERNAL IMAGE PREDICTION")
print("============================================")

# Example:
# If you have a 28x28 grayscale Fashion-MNIST image,
# load it and convert it into a 784-feature vector.

# Example using one test image as external input

external_image = X_test[0]

external_image = external_image.reshape(1, 784)


# Predict probability

external_probability = model.predict(
    external_image
)[0]


# Get predicted class

external_prediction = np.argmax(
    external_probability
)


print(
    "\nPredicted Class:",
    external_prediction
)

print(
    "Predicted Item:",
    class_names[external_prediction]
)

print(
    "Prediction Probability:",
    external_probability[external_prediction]
)


# ============================================================
# 27. DISPLAY EXTERNAL IMAGE
# ============================================================

plt.figure(figsize=(4, 4))

plt.imshow(
    X_test[0].reshape(28, 28),
    cmap="gray"
)

plt.title(
    f"Predicted: {class_names[external_prediction]}"
)

plt.axis("off")

plt.show()
```
